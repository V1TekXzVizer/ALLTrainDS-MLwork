```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as  sns
```

```python
df = pd.read_csv("ds_salaries.csv")
```

```python
df.head(5)
df.info()
df.describe()
```

```python
df.isnull().sum()
```

```python
df.describe(include=['object'])
```

```python
df['job_title'].value_counts().head(10)
df['experience_level'].value_counts()
df['employment_type'].value_counts()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(df['salary_in_usd'], bins=30)
plt.title('Wyplata w USD')
plt.show()
top_jobs = df['job_title'].value_counts().head(10).index
top_df = df[df['job_title'].isin(top_jobs)]
plt.figure(figsize=(12,6))
sns.boxplot(x='job_title', y='salary_in_usd', data=top_df)
plt.xticks(rotation=45)
plt.title("Wyplata wedlug miejsca")
plt.show()
```

```python
numeric_cols = df.select_dtypes(include=['number'])
plt.figure(figsize=(10,8))
sns.heatmap(numeric_cols.corr(), annot=True, cmap='coolwarm')
plt.title('Korelacja miedzy cechami')
plt.show()
```

```python
df.groupby('experience_level')['salary_in_usd'].mean().sort_values(ascending=False)
df.groupby('company_size')['salary_in_usd'].mean().sort_values(ascending=False)
df.groupby('company_location')['salary_in_usd'].mean().sort_values(ascending=False).head(10)
```
